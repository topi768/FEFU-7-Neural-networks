# 1. Выбор модели

Для суммаризации диалога выбрана языковая модель `xiaomi/mimo-v2.5`, доступная через API Cloud.ru. Модель будет получать текст расшифровки и формировать конспект. Генерация выполняется в облаке

```python
load_dotenv(".env")
```

```python
import os

from openai import APIStatusError, OpenAI

try:
    with OpenAI(
        api_key=os.environ["API_KEY"],
        base_url="https://foundation-models.api.cloud.ru/v1",
    ) as client:
        model_ids = [model.id for model in client.models.list()]
        print("Моделей в списке:", len(model_ids))
        print("MiMo есть в списке:", "xiaomi/mimo-v2.5" in model_ids)
except APIStatusError as e:
    print("Статус:", e.status_code)
    print("Ответ сервера:", repr(e.response.text))
```

```python
for model_id in model_ids:
    if model_id.lower().startswith(("xiaomi/", "qwen/")):
        print(model_id)
```

```python
print([model_id for model_id in model_ids if "mimo" in model_id.lower()])
```
